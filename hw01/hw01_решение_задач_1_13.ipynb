````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
cities_clean = [i.lower().title().replace(" ", "") for i in cities]
print(cities_clean)
````

````python
amounts_int = [int(i) for i in amounts]
print(amounts_int)
````

````python
print(len(amounts)) # число заказов
print(sum(amounts_int)) # общая сумма
print(round(sum(amounts_int) / len(amounts))) # среднее округленное до целого
````

````python
not_paid = []
index_cnt = 0
for i in paid:
  if not i:
    not_paid.append(index_cnt)
  index_cnt += 1
_sum = sum(amounts_int[i] for i in not_paid)
print(f"Количество неоплаченных заказов = {len(not_paid)}, на сумму = {_sum}")
````

````python
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    status = "Оплачен" if is_paid else "Не оплачен"
    print(f"{city} | {amount} | {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
almaty_revenue = 0
astana_revenue = 0

for i, j, k in zip(cities_clean, amounts_int, paid):
    if i == "Almaty" and k:
        almaty_revenue += j

    elif i == "Astana" and k:
        astana_revenue += j

print(f"Выручка в Алмате: {almaty_revenue}")
print(f"Выручка в Астане: {astana_revenue}")
````

````python
paid_amounts = [amount if is_paid else -1
                for amount, is_paid in zip(amounts_int, paid)]

a = paid_amounts.index(max(paid_amounts))

print(f"Номер заказа: {a + 1}")
print(f"Город: {cities_clean[a]}")
````

````python
выходные = []
for i, j in enumerate(days):
  if j in ("сб", "вс"):
    выходные.append(i)
weekend_share = len(выходные) / len(days) * 100 # проценты
print(f"{weekend_share:.1f}%")
````

````python
# задача копилка (1)
student_invest = 15000
invested = 0
months1 = 0

while invested < 400000:
  months1 += 1
  invested += student_invest

print(months1)

# задача, если откладывать +1000 каждый месяц
student_invest = 15000
invested = 0
months2 = 0

while invested < 400000:
  months2 += 1
  invested += student_invest
  student_invest += 1000

print(months2)
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
unique_cities = list(set(cities_clean))
revenue_by_city = [[city, 0] for city in unique_cities]

for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid:
        for row in revenue_by_city:
            if row[0] == city:
                row[1] += amount
                break

revenue_by_city.sort(key=lambda row: row[1], reverse=True)
unique_cities = [row[0] for row in revenue_by_city]

print("Cities:", unique_cities)
print("Revenue:", revenue_by_city)
````

````python
sorted_amounts = sorted(amounts_int)
n = len(sorted_amounts)

if n == 0:
    print("No orders")
elif n % 2 == 1:
    median = sorted_amounts[n // 2]
    print("Median:", median)
else:
    median = (sorted_amounts[n // 2 - 1] + sorted_amounts[n // 2]) / 2
    print("Median:", median)
````

````python
weekdays = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]
daily_revenue = [0] * 7

for day, amount, is_paid in zip(days, amounts_int, paid):
    if is_paid:
        index = weekdays.index(day)
        daily_revenue[index] += amount

for day, revenue in zip(weekdays, daily_revenue):
    bars = "█" * (revenue // 5000)
    print(f"{day} | {bars} {revenue}")
````

В данных были некоторые проблемы с форматированием. Например, названия городов могли отличаться из-за лишних пробелов или регистра, хотя это один и тот же город. Также некоторые суммы заказов были записаны как строки, а не числа, поэтому их пришлось конвертировать в int. В общем, перед анализом данные нужно сначала привести к нормальному виду, иначе результаты могут быть неправильными.
